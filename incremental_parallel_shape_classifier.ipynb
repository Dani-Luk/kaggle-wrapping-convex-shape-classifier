### 🧩 Key Features:
- **Incremental Block Learning**: Dynamically adds polygon blocks to handle misclassified clusters
- **Parallel Architecture**: All blocks processed in parallel, combined via trainable Dense layer
- **Geometric Initialization**: Uses circle/polygon approximations for better starting points
- **Cluster Detection**: Automatically finds FP/FN clusters using DBSCAN
- **Smart Weight Initialization**: Heuristic initialization of combiner weights based on cluster samples
- **Rich Visualizations** – provides clear insights into block hyperplanes and resulting decision boundaries.

### 🏗️ Architecture Overview

```
                  ┌──▶ [PolygonBlock₁] ──┐
Input(2D) ────────┼──▶ [PolygonBlock₂] ──┼──▶ Concatenate ─▶ Dense(1, no bias) ─▶ Sigmoid ─▶ Output
                  │         ...          │
                  └──▶ [PolygonBlockₘ] ──┘
```

Each **PolygonBlock** contains:
- A **Dense** layer with ```n_sides``` neurons — each defines a hyperplane; their intersection forms the polygon interior.
- A **ReLU** activation ```h = max(0, s)``` → selects the positive side of each hyperplane.
- A **product aggregation** (logical AND): ```z = ∏ h``` → intersection corresponds to the polygon interior.
- **Trainable sharpness** (scaling) and **bias** parameters: ```output = K * z + b```.

### 🔁Workflow:
1. Initialize the first block as an _n_-sided regular polygon with corresponding  ```W_INIT``` positive weight, covering the entire shape.
2. Train the initial model
3. Find largest error cluster (FP or FN)
4. Add new block with heuristic weight initialization around that cluster
5. Train all layers with boosted sample weights on cluster points and a **MultiOptimizer** that favors block-line movement over sharpness.
6. Repeat (from step 3) until no significant clusters remain or **max_blocks** reached
7. Fine-tune all layers

### Advantages of Parallel Architecture:
- **Learned combination of blocks** — flexible weighting instead of fixed additive logic.
- **Improved capacity** to model complex or highly non-convex shapes.
- **Independent contribution** — each block influences the final decision separately.

### Utility Functions

````python
def compute_iou(y_true, y_pred_probs, threshold=0.5):
    """Compute Intersection over Union (IoU) for binary classification."""
    y_pred = (y_pred_probs > threshold).astype(int)
    intersection = np.sum((y_true == 1) & (y_pred == 1))
    union = np.sum((y_true == 1) | (y_pred == 1))
    return intersection / (union + 1e-8)

print_verbose("✓ Utility functions ready")
````

## 3. Data Generation

We need a diverse set of 2-D shapes to test our incremental learning approach.  
Our data generators produce a variety of challenging geometries — from simple convex forms to complex non-convex and holed shapes.

- **Circles / Ellipses** – test basic convex approximation
- **L-Shapes / S-Shapes** – test non-convex region handling
- **Stars** – test highly concave boundaries
- **Donuts** – test holes and exclusion regions
- **Triangle-with-hole** – test nested boundaries and shape subtraction
- **Figure-8 shapes** – test complex real-world topologies

````python
def generate_disk_data(random_seed=None, n_samples=3000, radius_x=1.0, radius_y=1.0):
    """
    Generate binary classification data for ellipse/circle shapes.
    
    Args:
        n_samples: Number of random points to generate
        radius_x: Semi-axis length in x direction
        radius_y: Semi-axis length in y direction
        
    Returns:
        X: Array of 2D points
        y: Binary labels (1=inside, 0=outside)
    """
    # Set random seed
    if random_seed is None:
        random_seed = random.randint(1, 1000)
    set_random_seed(random_seed)
    
    X = np.random.uniform(-2.5, 2.5, size=(n_samples, 2))
    
    # Ellipse equation: (x/a)² + (y/b)² < 1
    inside_ellipse = (X[:, 0]**2 / radius_x**2 + X[:, 1]**2 / radius_y**2) < 1
    y = inside_ellipse.astype(np.float32)
    
    return X, y


def generate_polygon_data(random_seed=None, corners=[(0, 1.5), (-1.3, -0.75), (1.3, -0.75)], n_samples=3000 ):
    """
    Generate binary classification data for N-sided polygons using ray casting.
    Args:
        n_samples: Number of random points to generate
        corners: List of N vertices in counter-clockwise order (Default: Equilateral triangle)
            Returns:
        X: Array of 2D points
        y: Binary labels (1=inside, 0=outside)
    """
    # Set random seed
    if random_seed is None:
        random_seed = random.randint(1, 1000)
    set_random_seed(random_seed)
    
    X = np.random.uniform(-3, 3, size=(n_samples, 2))
    
    def point_in_polygon(point, polygon_corners):
        """Ray casting algorithm for point-in-polygon test"""
        x, y = point
        n = len(polygon_corners)
        inside = False
        
        p1x, p1y = polygon_corners[0]
        for i in range(1, n + 1):
            p2x, p2y = polygon_corners[i % n]
            if y > min(p1y, p2y):
                if y <= max(p1y, p2y):
                    if x <= max(p1x, p2x):
                        if p1y != p2y:
                            xinters = (y - p1y) * (p2x - p1x) / (p2y - p1y) + p1x
                        if p1x == p2x or x <= xinters:
                            inside = not inside
            p1x, p1y = p2x, p2y
        return inside
    
    y = np.array([point_in_polygon(point, corners) for point in X]).astype(np.float32)
    return X, y


def generate_donut_data(random_seed=None, n_samples=3000, inner_radius=1.00, outer_radius=2.40):
    """Generate donut/annulus shape data centered in origin."""
    # Set random seed
    if random_seed is None:
        random_seed = random.randint(1, 1000)
    set_random_seed(random_seed)

    X = np.random.uniform(-3, 3, (n_samples, 2))
    dist = np.sqrt((X[:,0])**2 + (X[:,1])**2)
    y = ((dist < outer_radius) & (dist > inner_radius)).astype(np.float32)
    return X, y


def generate_eight_shape_data_normalized(random_seed=None, n_samples=3000, inner_radius_1=0.11, outer_radius_1=0.25, inner_radius_2=0.14, outer_radius_2=0.28):
    """Generate figure-eight or double donut shape data."""
        # Set random seed
    if random_seed is None:
        random_seed = random.randint(1, 1000)
    set_random_seed(random_seed)
    X = np.random.uniform(0, 1, (n_samples, 2))
    cx_1, cy_1 = np.random.uniform(0.3, 0.35, 2)
    cx_2, cy_2 = np.random.uniform(0.6, 0.65, 2)
    dist_1 = np.sqrt((X[:,0]-cx_1)**2 + (X[:,1]-cy_1)**2)
    dist_2 = np.sqrt((X[:,0]-cx_2)**2 + (X[:,1]-cy_2)**2)
    y = (((dist_1 < outer_radius_1) & (dist_1 > inner_radius_1)) | ((dist_2 < outer_radius_2) & (dist_2 > inner_radius_2))).astype(np.float32)
    return X, y 


def generate_n_star_data(random_seed=None, n_samples=3000, n_corners=5, inner_radius=1.25, outer_radius=2.50):
    """Create n-pointed star shape data by defining outer and inner corners.
    Args:
        n: Number of data points
        n_corners: Number of star points
        inner_radius: Radius of inner vertices
        outer_radius: Radius of outer vertices (star tips)
    """
        # Set random seed
    if random_seed is None:
        random_seed = random.randint(1, 1000)
    set_random_seed(random_seed)
    angles = np.linspace(0, 2 * np.pi, n_corners, endpoint=False)
    # Adding some noise to make it less regular
    angles += np.random.uniform(-0.1, 0.1, size=n_corners)
    outer_corners = np.c_[np.cos(angles), np.sin(angles)] * outer_radius
    inner_corners = np.c_[np.cos(angles + np.pi / n_corners), np.sin(angles + np.pi / n_corners)] * inner_radius
    corners = []
    for i in range(n_corners):
        corners.append((outer_corners[i,0], outer_corners[i,1]))
        corners.append((inner_corners[i,0], inner_corners[i,1]))
    return generate_polygon_data(corners=corners, n_samples=n_samples, random_seed=random_seed)

print_verbose("✓ Data generation functions ready")

````

## 4. Visualization Tools

To better understand how the incremental approach behaves, we provide visualization utilities that display:

- **Decision boundaries** – regions where the model separates classes
- **Hyperplane lines** – edges formed by each polygon block
- **True boundaries** – ground-truth shape contours for comparison
- **Classification results** – correctly and incorrectly classified points

````python
class TrueBoundaryApprox:
    """
    Creates a smooth approximation of the true shape boundary for visualization.

    This class precomputes and stores a high-resolution grid, then finds the nearest
    labeled point for each grid location to generate smooth boundary curves for future plots.
    """
        
    def __init__(self, X, y, x_range=(-3, 3), resolution=60, color='yellow', alpha=0.8):
        self.X = X
        self.y = y
        self.x_range = x_range
        self.resolution = resolution
        self.color = color
        self.alpha = alpha

        # Precompute grid and labels (expensive part)
        self._compute_grid_labels()

    def _compute_grid_labels(self):
        """Pre-compute grid labels using nearest neighbor approximation."""
        xx, yy = np.meshgrid(
            np.linspace(*self.x_range, self.resolution),
            np.linspace(*self.x_range, self.resolution)
        )
        grid_points = np.c_[xx.ravel(), yy.ravel()]
        distances = cdist(grid_points, self.X)
        closest_indices = np.argmin(distances, axis=1)
        grid_true_labels = self.y[closest_indices].reshape(xx.shape)
        self.xx, self.yy, self.grid_true_labels = xx, yy, grid_true_labels

    def add_to_plot(self, ax):
        """Add true boundary contour to existing plot."""
        ax.contour(
            self.xx, self.yy, self.grid_true_labels,
            levels=[0.5], # Decision boundary
            colors=[self.color],
            linewidths=3,
            linestyles='--',
            alpha=self.alpha
        )

def plot_dataset_predictions_and_boundaries(model, X, y_true, ax, x_range=(-3, 3), resolution=100):
    """
    Plot dataset with model predictions and boundaries by showing correct/incorrect classifications to multiple axes
    """
    # Get model predictions (calculate once, reuse for all axes)
    y_pred = model.predict(X, verbose=0).flatten()
    pred_labels = (y_pred > 0.5).astype(int)
    
    # Create background grid for prediction visualization with configurable resolution
    xx, yy = np.meshgrid(np.linspace(*x_range, resolution), np.linspace(*x_range, resolution))
    grid_points = np.c_[xx.ravel(), yy.ravel()]
    grid_pred = model.predict(grid_points, verbose=0).reshape(xx.shape)

    # Plot prediction background (decision regions)
    ax.contourf(xx, yy, grid_pred, levels=5, cmap="RdBu", alpha=0.6, vmin=0, vmax=1)

    # Plot decision boundary
    ax.contour(xx, yy, grid_pred, levels=[0.5], colors="black", linewidths=3, alpha=0.9)

    # Plot points with correct/incorrect markers
    for i in range(len(X)):
        is_correct = (pred_labels[i] == y_true[i])
        marker = 'o' if is_correct else 'x'
        size = 25 if is_correct else 35
        color = 'red' if pred_labels[i] == 1 else 'blue'
        alpha = 0.7 if is_correct else 0.9
        edge_width = 0.5 if is_correct else 1.2
        ax.scatter(X[i, 0], X[i, 1], c=color, marker=marker, s=size,
                   alpha=alpha, edgecolors='black', linewidth=edge_width)
# end plot_dataset_predictions_and_boundaries


def plot_layer(ax, block_layer, x_range=(-3, 3), sign=+1, alpha=0.8):
    """
    Plot the model weight lines(polygon lines) into ax axis, color-coded by block_layer.alpha sign:
        - Red lines: sign positive (additive regions)
        - Blue lines: sign negative (subtractive regions)
    """
    weights, biases = block_layer.poly.get_weights()
    # decide color by sign
    color = "red" if sign >= 0 else "blue"

    for i in range(weights.shape[1]):
        w1, w2 = weights[:, i]
        bias = biases[i]
        
        if abs(w2) > 1e-8:  # Non-vertical line: w1*x + w2*y + bias = 0
            # Use two points to define the line within the x_range
            x1, x2 = x_range[0], x_range[1]
            y1 = -(w1 * x1 + bias) / w2
            y2 = -(w1 * x2 + bias) / w2
            # Use axline with two points and clip to axes limits
            ax.axline((x1, y1), (x2, y2), color=color, linewidth=1.8, alpha=alpha)
        else:  # Vertical line: x = -bias/w1
            x_line = -bias / w1
            # Only draw if within x_range
            if x_range[0] <= x_line <= x_range[1]:
                ax.axvline(x=x_line, color=color, linewidth=1.8, alpha=alpha)
    
    # Set axis limits to ensure clipping
    ax.set_xlim(x_range)
    ax.set_ylim(x_range)
    return
# end plot_layer


def plot_all_blocks(model, ax, x_range=(-3, 3)):
    """Plot all PolygonBlocks in the model."""
    block_layers = [l for l in model.layers if "polygon_block" in l.name]
    n_blocks = len(block_layers)
    w = model.get_layer("dense_combiner").get_weights()[0]

    for i, block in enumerate(block_layers):
        plot_layer(ax, block, x_range, w[i], alpha = 0.5 if i < n_blocks - 1 else 0.8)  # last block more opaque


def plot_model(plot_verbose, model, X_train, y_train, X_val, y_val, title="", true_boundary_approx:TrueBoundaryApprox|None=None):
    """
    Create comprehensive training result visualization.
    - Training data points
    - True boundary approximation (if provided)
    - Model decision boundary and polygon hyperplanes regions
    - Performance metrics in title
    """    
    if plot_verbose:
        ax = plt.gca()
        ax.set_aspect('equal')
        ax.grid(True, alpha=0.3)        
        plot_dataset_predictions_and_boundaries(model, X_train, y_train, ax)
        plot_all_blocks(model, ax)
        if true_boundary_approx:
            true_boundary_approx.add_to_plot(ax)
        val_iou = compute_iou(y_val, model.predict(X_val, verbose=0).flatten())
        train_iou = compute_iou(y_train, model.predict(X_train, verbose=0).flatten())
        ax.set_title(f"{title}: {train_iou=:.2%}, {val_iou=:.2%} ", fontsize=16)
        plt.show()

print_verbose("✓ Visualization functions ready")
````

## 5. Geometric Weight Initialization

Instead of random initialization, we use **geometric intuition** to provide the model with a strong starting point.

Each polygon block represents a set of hyperplanes. By initializing these hyperplanes to approximate a circle (an n-sided regular polygon), the model begins with a balanced and spatially uniform configuration for discovering cluster boundaries.

````python
def get_polygon_circle_weights(n_sides=4, radius=2.0):
    """
    Initialize polygon layer weights to approximate a circle.
    The catches: 
        - keep left! ;) when generating the polygon so the inside and outside will be well classified  
        - choose a radius large enough to contain the shape, 
        and don't forget to leave enough points outside, to have thereby good examples outside 
        and so the model will be able to start the learning process

    This geometric initialization ensures the model starts with a reasonable
    approximation rather than random weights.
    
    Args:
        n_sides: Number of polygon sides (hyperplanes)
        radius: Initial radius of the approximated circle
        
    Returns:
        weights: Weight matrix [2, n_sides] 
        biases: Bias vector [n_sides]
    """
    assert n_sides > 2, "Number of sides must be greater than 2 anyway"
    angles = np.linspace(0, 2 * np.pi, n_sides, endpoint=False)
    
    # Each column represents a hyperplane normal vector
    # Use outward-pointing normals: for w·x + b > 0 to define "inside"
    # For circle: w·x + b = cos(θ)·x + sin(θ)·y + R = 0 is the boundary
    weights = np.stack([np.cos(angles), np.sin(angles)], axis=1).T / radius
    biases = np.ones(n_sides)  # This gives b = 1, so boundary is at radius = 1*radius = radius

    return weights, biases

def get_cluster_weights(X_cluster, cluster_type, n_sides):
    """
    Return weights&biases for a polygon (triangle by default) to enclose a cluster.

    Args:
        X_cluster: (N, 2) array of points in the selected cluster
        n_sides: number of polygon sides (3 for triangle)

    Returns:
        weights: (2, n_sides)
        biases: (n_sides,)
    """
    assert X_cluster.shape[1] == 2, "Cluster points must be 2D"

    # 1) Find bounding box
    x_min, y_min = np.min(X_cluster, axis=0)
    x_max, y_max = np.max(X_cluster, axis=0)

    # 2) Compute center and "radius" (half of diagonal)
    cx, cy = (x_min + x_max) / 2, (y_min + y_max) / 2

    if cluster_type == 'FN':
        # For False Negative clusters (inside points), we'll try IN-OUT(expanding) so starting from min radius = inscribed circle into the rectangle
        radius = min((x_max - x_min), (y_max - y_min)) / 2.0  # aprox inscribed circle radius
    else:
        # For False Positive clusters (outside points), we'll try OUT-IN(contracting) so starting from max radius = half diagonal of bounding box
        radius = max((x_max - x_min), (y_max - y_min)) / 2.0  # aprox circumscribed circle radius

    print_verbose(f"Estimated: cluster center(x, y) = ({cx:.2f}, {cy:.2f}), radius = {radius:.2f}")

    # 3) Reuse circle initializer with translation
    weights, biases = get_polygon_circle_weights(n_sides=n_sides, radius=radius)

    # 4) Adjust biases to shift polygon to cluster center
    # Each hyperplane is defined by w·x + b > 0 for "inside"
    # To translate: replace b with b - w·center (so w·(x-center) + b > 0 becomes w·x + (b - w·center) > 0)
    # NOTE: weights shape is (2, n_sides) → each column = normal vector

    translations = np.dot(weights.T, np.array([cx, cy]))
    biases = biases - translations  # very important: subtract the translation! because of the sign convention in get_polygon_circle_weights()

    return weights, biases

print_verbose("✓ Weight initialization functions ready")
````

## 6. Cluster Detection

The strategic placement of new blocks is guided by density-based clustering (DBSCAN), which identifies the largest false-positive or false-negative cluster to target next.

````python
def find_largest_error_cluster(X, y_true, y_pred, eps=CLUSTER_EPS, min_samples=CLUSTER_MIN_SAMPLES):
    """
    Find the largest FP or FN cluster separately.

    Args:
        X: (N, 2) input points
        y_true: (N,) ground-truth labels (0/1)
        y_pred: (N,) model predictions (0-1)
        eps: DBSCAN neighborhood radius
        min_samples: DBSCAN minimum points per cluster

    Returns:
        cluster_type: "FP" or "FN"
        X_cluster: points in largest cluster
        idx_cluster: indices of points in largest cluster
    """
    # 1) Split errors
    FP_idx = np.where((y_true == 0) & (y_pred > 0.5))[0]
    FN_idx = np.where((y_true == 1) & (y_pred <= 0.5))[0]

    # Helper to run DBSCAN and find largest cluster
    def run_dbscan(indices, label):
        if len(indices) == 0:
            return None, None, None
        X_err = X[indices]
        db = DBSCAN(eps=eps, min_samples=min_samples).fit(X_err)
        labels = db.labels_

        # Ignore noise points (-1)
        clusters = [np.where(labels == c)[0] for c in set(labels) if c != -1]
        if not clusters:
            return None, None, None

        # Pick largest cluster
        largest = max(clusters, key=len)
        return label, X_err[largest], indices[largest]

    # 2) Run separately for FP and FN
    cluster_fp = run_dbscan(FP_idx, "FP")
    cluster_fn = run_dbscan(FN_idx, "FN")

    # 3) Choose the larger cluster (if both exist)
    if cluster_fp[1] is None and cluster_fn[1] is None:
        return None, None, None
    if cluster_fp[1] is None: return cluster_fn
    if cluster_fn[1] is None: return cluster_fp

    return cluster_fp if len(cluster_fp[1]) >= len(cluster_fn[1]) else cluster_fn # NOTE: what about ">=" vs ">" ...?

print_verbose("✓ Cluster detection function ready")
````

## 7. Polygon Block Architecture

The **PolygonBlock** serves as the fundamental unit of our approach, with each block defining a polygon that either adds to or subtracts from the overall decision boundary.

### Mathematical Foundation:
- Each layer has `n_sides` hyperplanes: `w_i · x + b_i`
- ReLU activation: `h_i = max(0, w_i · x + b_i)`
- Product operation: `z = ∏ h_i` (AND logic - all hyperplanes must be satisfied)
- Final contribution: `K * z + b` -> !!! no α, no sigmoid here (the final Dense weights will control this)

````python
class PolygonBlock(tf.keras.layers.Layer):
    """
    A parallel polygon block that outputs K * z_in + b (no alpha, no sigmoid).
    
    Key differences from sequential PolygonLayer:
    - No alpha parameter (will be in final Dense layer)
    - No sigmoid (applied globally at the end)
    - No z_old input (blocks run in parallel)
    
    Mathematical flow:
    1. Linear: s = W·x + b  (hyperplanes)
    2. ReLU: h = max(0, s)  (positive side of each hyperplane)
    3. Product: z = ∏ h     (intersection = polygon interior)
    4. Scale: output = K * z + b  (trainable sharpness and bias)
    """

    def __init__(self, weights, biases, n_sides=3, name="polygon_block", **kwargs):
        super().__init__(name=name, **kwargs)
        self.n_sides = n_sides
        self.init_weights = (weights, biases)
        # define Dense as a sub-layer, Keras will manage its kernel+bias
        self.poly = Dense(self.n_sides, use_bias=True, activation=None, name=f"{name}_s")

    def build(self, input_shape):
        """Initialize all trainable parameters."""
        
        with tf.name_scope(self.name):
            # initialize the dense layer with our geometric weights
            self.poly.build((None, 2))

            self.poly.set_weights(self.init_weights)
                # rename the inner Dense weights so we can find them later
            self.poly.kernel._handle_name = f"{self.name}_poly_kernel"
            self.poly.bias._handle_name   = f"{self.name}_poly_bias"

            # trainable sharpness parameter K (controls polygon edge steepness)
            self.K = self.add_weight(
                name=f"{self.name}_sharpness",
                shape=(),
                initializer=tf.keras.initializers.Constant(10),
                trainable=True
            )

            # trainable bias term to shift polygon contribution
            self.b = self.add_weight(
                name=f"{self.name}_bias",
                shape=(),
                initializer="zeros",
                trainable=True
            )

    def call(self, xy):
        """Forward pass through the polygon block."""
        # Compute polygon activations
        s = self.poly(xy)                   # half-space values
        h_in = tf.nn.relu(s)                 # keep inside-edge activations
        z_in = tf.reduce_prod(h_in, axis=-1, keepdims=True)  # AND over edges
        
        # Return scaled output (no sigmoid here)
        return self.K * z_in + self.b
    
print_verbose("✓ PolygonBlock layer defined")
````

## 8. Parallel Model Assembly

````python
class MultiOptimizer(tf.keras.optimizers.Optimizer):
    """Optimizer that combines multiple sub-optimizers. Used to prioritize training of block lines over sharpness."""
    def __init__(self, optimizers_and_vars, name="MultiOptimizer", **kwargs):
        super().__init__(name=name, learning_rate=0.0, **kwargs)
        self.optimizers_and_vars = []
        for opt, vars_ in optimizers_and_vars:
            names = [v.name for v in vars_]
            self.optimizers_and_vars.append((opt, vars_, names))

    def apply_gradients(self, grads_and_vars, **kwargs):
        grads_and_vars = list(grads_and_vars)   # 👈 materialize the zip iterator
        if INSPECT:
            print(f"Optimizer called for {len(grads_and_vars)} grads_and_vars, {kwargs=}")

        for opt, vars_, names in self.optimizers_and_vars:
            sub = [(g, v) for g, v in grads_and_vars if v.name in names and g is not None]
            if sub:
                if INSPECT:
                    print(f"Optimizer {opt.name} with lr={float(tf.keras.backend.get_value(opt.learning_rate))} "
                        f"applying to {[getattr(v, '_handle_name', v.name) for _, v in sub]}"
                        )
                opt.apply_gradients(sub, **kwargs)

# hybrid loss function combining binary cross-entropy and mean squared error
def hybrid_loss(y_true, y_pred, alpha=0.1):
    bce = tf.keras.losses.binary_crossentropy(y_true, y_pred)
    mse = tf.square(y_true - y_pred)
    return (1 - alpha) * bce + alpha * mse


def build_initial_model(weights, biases, n_sides=3):
    """
    Build the initial parallel model with one polygon block.
    
    Architecture:
    - Input: 2D coordinates (x, y)
    - PolygonBlock: First shape approximation
    - Dense(1): Single weight (initialized to W_INIT)
    - Sigmoid: Global activation with K_SIGMOID sharpness
    
    Args:
        weights, biases: Geometric initialization for hyperplanes
        n_sides: Number of polygon sides
        
    Returns:
        Compiled Keras model ready for training
    """
    xy = Input(shape=(2,), name="xy")
    
    # First polygon block
    block = PolygonBlock(weights, biases, n_sides=n_sides, name="polygon_block_0")
    z1 = block(xy)
    
    # Dense layer with a first single weight (initialized to W_INIT)
    # NOTE: use_bias=False to avoid shifting the scaled outputs of polygon blocks

    combined = Dense(1, use_bias=False, 
                     kernel_initializer=tf.keras.initializers.Constant(W_INIT),
                     name="dense_combiner")(z1)
    
    # Global sigmoid activation
    out = Activation(lambda x: tf.sigmoid((K_SIGMOID) * x), name="out")(combined)
    
    model = Model(inputs=xy, outputs=out)

    opt_poly = tf.keras.optimizers.Adam(2e-3)
    opt_rest = tf.keras.optimizers.Adam(4e-4)    

    poly_vars = [
        v for v in model.trainable_variables
        if hasattr(v, "_handle_name") and "_poly_" in v._handle_name
    ]
    poly_names = {getattr(v, "_handle_name", v.name) for v in poly_vars}

    rest_vars = [
        v for v in model.trainable_variables
        if getattr(v, "_handle_name", v.name) not in poly_names
    ]
    if INSPECT:
        intersect = set(getattr(v, "_handle_name", v.name) for v in poly_vars) & \
                set(getattr(v, "_handle_name", v.name) for v in rest_vars)
        print("Overlap:", intersect)

        print("Polygon variables:")
        for v in poly_vars:
            print(v._handle_name if hasattr(v, "_handle_name") else "N/A", "→", v.name)
        print("Rest variables:")
        for v in rest_vars:
            print(v._handle_name if hasattr(v, "_handle_name") else "N/A", "→", v.name)

    optimizer = MultiOptimizer([
        (opt_poly, poly_vars),
        (opt_rest, rest_vars),
    ])
    model.compile(optimizer=optimizer, loss="mse", metrics=[tf.keras.metrics.BinaryAccuracy()])
    
    if INSPECT:
        print("poly vars:", len(poly_vars))
        print("rest vars:", len(rest_vars))
        print("opt_poly LR =", opt_poly.learning_rate.numpy(),
            "| opt_rest LR =", opt_rest.learning_rate.numpy())

    return model


def estimate_new_block_weight_old(model, X_cluster, cluster_type):
    """
    Estimate the initial weight for a new block based on current model errors.
    
    Strategy:
    - Get model output before sigmoid (from dense_combiner layer)
    - Calculate average error magnitude on cluster points
    - FN clusters: return +|error| (need to add positive contribution)
    - FP clusters: return -|error| (need to subtract to cancel false positives)
    
    Args:
        model: Current model
        X_cluster: Points in the error cluster
        cluster_type: 'FN' or 'FP'
        
    Returns:
        weight_init: Initial weight for new block
    """
    # Get model output before sigmoid
    dense_layer = model.get_layer('dense_combiner')
    pre_sigmoid_model = Model(inputs=model.input, outputs=dense_layer.output)
    
    # Calculate pre-sigmoid values on cluster
    pre_sigmoid_vals = pre_sigmoid_model.predict(X_cluster, verbose=0).flatten()
    
    # Average absolute value (magnitude of error)
    # avg_magnitude = np.mean(np.abs(pre_sigmoid_vals))
    avg_magnitude = np.mean(pre_sigmoid_vals)
    if INSPECT:
        min_magnitude = np.min(pre_sigmoid_vals)
        max_magnitude = np.max(pre_sigmoid_vals)
        print(f"pre_sigmoid_vals: min={min_magnitude:.3f}, max={max_magnitude:.3f}, mean={avg_magnitude:.3f}")
    avg_magnitude = np.abs(avg_magnitude)
    
    # Ensure we have a reasonable minimum magnitude
    avg_magnitude = max(avg_magnitude, 0.5) / 2
    
    # Sign depends on cluster type
    if cluster_type == 'FN':
        # False Negatives: need to add positive contribution
        weight_init = +avg_magnitude
    else:
        # False Positives: need to subtract
        weight_init = -avg_magnitude
    
    print_verbose(f"Estimated weight for new block: {weight_init:.3f} (cluster type: {cluster_type})")
    
    return weight_init


def estimate_new_block_weight(model, X_cluster, cluster_type, weights, biases, n_sides):
    """
    Estimate initial dense_combiner weight for a new block,
    normalized by the block's own output magnitude.
    """
    # 1️⃣ Get pre-sigmoid outputs from current model
    dense_layer = model.get_layer('dense_combiner')
    pre_sigmoid_model = tf.keras.Model(inputs=model.input, outputs=dense_layer.output)
    pre_sigmoid_vals = pre_sigmoid_model.predict(X_cluster, verbose=0).flatten()

    # avg_pre_sigmoid = np.mean(np.abs(pre_sigmoid_vals))
    avg_pre_sigmoid = np.abs(np.mean(pre_sigmoid_vals))
    avg_pre_sigmoid = max(avg_pre_sigmoid, 0.5)  # floor

    # 2️⃣ Create temporary PolygonBlock (not part of model)
    temp_block = PolygonBlock(weights, biases, n_sides=n_sides)
    temp_block.build((None, 2))  # ensure internal Dense is built

    # 3️⃣ Compute mean raw output for this cluster
    block_out = temp_block(X_cluster, training=False).numpy().flatten()
    mean_block_val = np.mean(np.abs(block_out))
    mean_block_val = max(mean_block_val, 1e-5)  # avoid div-by-zero

    # 4️⃣ Combine magnitudes
    raw_weight = avg_pre_sigmoid / mean_block_val

    # 5️⃣ Sign depending on cluster type
    if cluster_type == 'FN':
        weight_init = +raw_weight
    else:
        weight_init = -raw_weight
    weight_init = weight_init / 1.05 # slight reduction to avoid overshooting
    print_verbose(
            f"Estimated combiner weight: {weight_init:.3f} "
            f"(cluster={cluster_type}, pre-sig={avg_pre_sigmoid:.3f}, block_out={mean_block_val:.3f})"
        )

    return weight_init

def add_parallel_block(model, weights, biases, n_sides=3, weight_init=1.0, name=None):
    """
    Add a new polygon block to the parallel architecture.
    
    Strategy:
    1. Extract outputs from all existing blocks
    2. Add new block in parallel 
    3. Concatenate all block outputs
    4. Create new Dense layer with all weights :old + new(weight_init)
    5. Apply global sigmoid
    
    Args:
        model: Existing trained model
        weights, biases: Initialization for new block
        n_sides: Number of sides for new polygon
        weight_init: Initial weight for new block in Dense layer
        name: Optional name for the new block
        
    Returns:
        New model with additional polygon block
    """
    
    xy = model.input
    
    # Get all existing polygon blocks
    existing_blocks = [l for l in model.layers if 'polygon_block' in l.name]
    n_existing = len(existing_blocks)
    
    # Get outputs from all existing blocks
    existing_outputs = [block.output for block in existing_blocks]
    
    # Create new block
    block_name = name or f"polygon_block_{n_existing}"
    new_block = PolygonBlock(weights, biases, n_sides=n_sides, name=block_name)
    new_output = new_block(xy)
    
    # Concatenate all block outputs
    all_outputs = existing_outputs + [new_output]
    concatenated = Concatenate(name=f"concat_{n_existing}")(all_outputs)
    
    # Get old weights from dense_combiner
    old_dense = model.get_layer('dense_combiner')
    old_weights = old_dense.get_weights()[0].flatten()  # Shape: (n_existing,)
    
    # Create new weight matrix: [old_weights..., weight_init]
    new_weights = np.append(old_weights, weight_init).reshape(-1, 1)
    
    # New Dense layer - create it first, then set weights
    dense_layer = Dense(1, use_bias=False, name="dense_combiner")
    combined = dense_layer(concatenated)
    
    # Now set the weights (must be done after the layer is built)
    dense_layer.set_weights([new_weights])
    
    # Global sigmoid
    out = Activation(lambda x: tf.sigmoid(K_SIGMOID * x), name="out")(combined)
    
    model_new = Model(inputs=xy, outputs=out)

    opt_poly = tf.keras.optimizers.Adam(3e-3)
    opt_rest = tf.keras.optimizers.Adam(5e-4)    
    poly_vars = [
        v for v in model.trainable_variables
        if hasattr(v, "_handle_name") and "_poly_" in v._handle_name
    ]
    poly_names = {getattr(v, "_handle_name", v.name) for v in poly_vars}
    
    rest_vars = [
        v for v in model.trainable_variables
        if getattr(v, "_handle_name", v.name) not in poly_names
    ]    
    if INSPECT:
        intersect = set(getattr(v, "_handle_name", v.name) for v in poly_vars) & \
                set(getattr(v, "_handle_name", v.name) for v in rest_vars)
        print("add_parallel_block Overlap:", intersect)

    optimizer = MultiOptimizer([
        (opt_poly, poly_vars),
        (opt_rest, rest_vars),
    ])
    if weight_init < 0: # for FP blocks use MSE loss ... :D
        model_new.compile(optimizer=optimizer, loss="mse", metrics=[tf.keras.metrics.BinaryAccuracy()])
    else:
        model_new.compile(optimizer=optimizer, loss="bce", metrics=[tf.keras.metrics.BinaryAccuracy()])

    return model_new


def validate_model_initialization(model, test_point=np.array([[0.0, 0.0]])):
    """
    Validate that the model initialization is correct by testing a point.
    
    The origin (0,0) should be classified as inside (class=1) for proper initialization.
    """
    prediction = model.predict(test_point, verbose=0)[0][0]
    
    try:
        assert prediction > 0.5, f"Test point prediction should be > 0.5, got {prediction:.4f}"
        print_verbose(f"Model initialization checked! (test point prediction = {prediction:.4f})")
        return True
    except AssertionError as e:
        print_verbose(f"Warning: {e}")
        return False

print_verbose("✓ Model building functions ready")
````

## 9. Main Training Algorithm

````python
def run_model_incremental(shape_type, X, y, 
                        max_blocks=3, n_sides=3, 
                        random_seed=None, n_samples=3000):
    """
    Incremental training with parallel polygon blocks.
    
    Training Workflow:
    1. Train initial shape - 50 epochs
    2. Evaluate → find FP/FN clusters
    3. Add new block in parallel with estimated weight
    4. Training all - 100 epochs
    5. Repeat until no clusters or max_blocks reached
    6. Final tuning of all layers - 50 epochs
    
    Args:
        shape_type: Name of the shape (for logging)
        X, y: Training data
        max_blocks: Maximum number of polygon blocks to add
        n_sides: Number of sides per polygon
        random_seed: Random seed for reproducibility
        n_samples: Number of samples (for logging)
        
    Returns:
        model: Trained model
        val_iou: Final validation IoU
        nb_blocks: Number of blocks in final model
    """
    # Set random seed
    if random_seed is None:
        random_seed = random.randint(1, 1000)
    set_random_seed(random_seed)
    
    print_verbose(f"{'='*60}")
    print_verbose(f"RUNNING Shape: {shape_type.upper()} with Parallel Incremental Blocks")
    print_verbose(f"Polygon sides per block: {n_sides}")
    print_verbose(f"Max blocks: {max_blocks}")
    print_verbose(f"{'='*60}")
    
    print_verbose(f"Inside points: {np.sum(y)}/{len(y)} ({100*np.sum(y)/len(y):.1f}%)")
    
    # Split data
    print_verbose("Splitting data.", end=' ')
    X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=VALIDATION_SIZE, random_state=random_seed)
   
    # Build and initialize model
    print_verbose("Building model.", end=' ')
    print_verbose("Initializing model weights.", end=' ')
    weights, biases = get_polygon_circle_weights(n_sides, radius=2.5)

    # ==================================================
    # Build initial model
    model = build_initial_model(weights, biases, n_sides)

    # Validate initialization
    validation_passed = validate_model_initialization(model)
    
    if not validation_passed:
        print_verbose("Model initialization validation failed - results may be poor")

    # Calculate initial IoU
    val_iou = compute_iou(y_val, model.predict(X_val, verbose=0).flatten())
    print_verbose(f"Initial validation IoU: {val_iou:.2%}")

    y_pred = model.predict(X_val, verbose=0).flatten()
    print_verbose(f"Fraction predicted positive: {np.mean(y_pred > 0.5):.2%}")

    true_boundary_approx = TrueBoundaryApprox(X, y, x_range=(-3, 3), resolution=100, color='yellow', alpha=0.8)
    
    # Plot initial state
    plot_model(True, model, X_train, y_train, X_val, y_val, 
               title=f"Initial", true_boundary_approx=true_boundary_approx)

    # =========================================
    # Initial training - 30 epochs
    print_verbose("Initial training (30 epochs)...")
    model.fit(X_train, y_train, epochs=30, batch_size=16, 
        class_weight={0: 1.0, 1: 2.5},
        verbose=0,
        validation_data=(X_val, y_val)
    )
    
    y_pred = model.predict(X_val, verbose=0).flatten()

    first_block = model.get_layer("polygon_block_0")
    print_block_Kb("After initial training first block", first_block)
    
    # Print Dense layer weights
    dense_weights = model.get_layer('dense_combiner').get_weights()[0].flatten()
    print_verbose(f"Dense weights: {dense_weights}")

    old_val_iou = val_iou = compute_iou(y_val, model.predict(X_val, verbose=0).flatten())
    train_iou = compute_iou(y_train, model.predict(X_train, verbose=0).flatten())
    print_verbose(f"After initial training IoU: train={train_iou:.2%}, val={val_iou:.2%}")
    
    # Plot after initial training
    plot_model(PLOT_VERBOSE, model, X_train, y_train, X_val, y_val, 
               title=f"After initial training", true_boundary_approx=true_boundary_approx)

    # Incrementally add blocks
    for i in range(max_blocks-1):
        print_verbose(f"{'-'*60}")
        print_verbose(f"Iteration {i+1}: Looking for error clusters...")
        
        # Evaluate on training set to find misclassified points
        y_train_pred = model.predict(X_train, verbose=0).flatten()
        cluster_type, X_cluster, idx_cluster = find_largest_error_cluster(X_train, y_train, y_train_pred)

        if X_cluster is not None:
            print_verbose(f"Found {cluster_type} cluster with {len(X_cluster)} points")
            
            # Initialize polygon around this cluster
            weights, biases = get_cluster_weights(X_cluster, cluster_type, n_sides)
            
            # Estimate weight for new block
            weight_init = estimate_new_block_weight(model, X_cluster, cluster_type, weights, biases, n_sides)
            
            # Add new block in parallel
            model = add_parallel_block(model, weights, biases, n_sides=n_sides, 
                                      weight_init=weight_init,
                                      name=f"polygon_block_{i+1}")
            
            new_block = model.get_layer(f"polygon_block_{i+1}")
            print_block_Kb(f"New block added: {new_block.name} with initial", new_block)
            
            # Print Dense layer weights
            dense_weights = model.get_layer('dense_combiner').get_weights()[0].flatten()
            print_verbose(f"Dense weights after adding block: {dense_weights}")

            plot_model(PLOT_VERBOSE, model, X_train, y_train, X_val, y_val, 
                      title=f"New block + {new_block.name}", true_boundary_approx=true_boundary_approx)

            sample_weight = np.ones(len(X_train))
            
            if weight_init > 0:
                sample_weight[idx_cluster] = 1.5  # Boost weight for cluster points
            else:
                sample_weight = np.where(y_train == 1, 2.0, 1.0) # {0:1.0, 1:2.0} 

            print_verbose(f"After training ALL blocks")
            model.fit(X_train, y_train, epochs=150, batch_size=32, 
                      verbose=0,
                    #   class_weight={0: 1.0, 1: 2.0},
                      sample_weight=sample_weight,
                      validation_data=(X_val, y_val),
                    #   callbacks=[early_stop]
                      )
            
            plot_model(PLOT_VERBOSE, model, X_train, y_train, X_val, y_val,
                      title=f"After training ALL blocks",
                      true_boundary_approx=true_boundary_approx)

            val_iou = compute_iou(y_val, model.predict(X_val, verbose=0).flatten())
            train_iou = compute_iou(y_train, model.predict(X_train, verbose=0).flatten())
            print_verbose(f"After block {new_block.name} added and trained: train={train_iou:.2%}, val={val_iou:.2%} (delta={val_iou - old_val_iou:+.2%})")
            
            # Print all block parameters
            for layer in model.layers:
                if "polygon_block" in layer.name:
                    print_block_Kb(f"  {layer.name}", layer)
            
            dense_weights = model.get_layer('dense_combiner').get_weights()[0].flatten()
            print_verbose(f"  Dense weights: {dense_weights}")
            
            old_val_iou = val_iou
            print_verbose(f"{'-'*60}\n")
        else:
            print_verbose(f"No significant clusters found (min_samples={CLUSTER_MIN_SAMPLES}) → stop adding layers")
            break
    else:
        print_verbose(f"Reached max blocks ({max_blocks}) → stop adding layers")
        
    # Final tuning of all layers
    print_verbose(f"{'-'*60}")
    print_verbose("Final tuning of all layers (50 epochs)...")
    
    model.compile(optimizer=tf.keras.optimizers.Adam(1e-4),
        loss="binary_crossentropy",
        metrics=[tf.keras.metrics.BinaryAccuracy()]
    )
    lr_scheduler = tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss", factor=0.5, patience=5, min_lr=1e-6
    )
    early_stop = tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True
    )
    model.fit(X_train, y_train, epochs=50, batch_size=64, verbose=0, 
              validation_data=(X_val, y_val),
              callbacks=[lr_scheduler, early_stop])

    train_iou = compute_iou(y_train, model.predict(X_train, verbose=0).flatten())
    val_iou = compute_iou(y_val, model.predict(X_val, verbose=0).flatten())
    
    print_verbose(f"{'='*60}")
    print_verbose(f"FINAL: train_IoU = {train_iou:.2%}, val_IoU = {val_iou:.2%}")
    print_verbose("-" * 60)

    nb_blocks = 0
    for layer in model.layers:
        if "polygon_block" in layer.name:
            print_block_Kb(f"  {layer.name}", layer)
            nb_blocks += 1
    
    dense_weights = model.get_layer('dense_combiner').get_weights()[0].flatten()
    print_verbose(f"  Dense weights: {dense_weights}")
    
    # Plot final model
    plot_model(True, model, X_train, y_train, X_val, y_val, 
               title=f"FINAL Model ({nb_blocks} blocks)", 
               true_boundary_approx=true_boundary_approx)

    return model, val_iou, nb_blocks

print_verbose("✓ Main incremental training function ready")
````

### 10.4 S-Shape _rotated_ by quadrilaterals

````python
# S-Shape rotated
random_seed = 42
n_samples = 3000
shape_name = 'S-shape'
corners = [
    (-2.0, -2.5),   # Start
    (2.0, -2.5),    # 4 right
    (2.0, 0.5),     # 2 up
    (-1.0, 0.5),    # 3.5 left
    (-1.0, 1.5),    # 1 up
    (2.0, 1.5),     # 3.5 right
    (2.0, 2.5),     # 0.5 up
    (-2.0, 2.5),    # 4 left
    (-2.0, -0.5),    # 2 down
    (1.0, -0.5),     # 3.5 right
    (1.0, -1.5),    # 1 down
    (-2.0, -1.5),   # 3.5 left
]
n_sides = 4 # quadrilateral blocks
max_blocks = 5

# convert s_shape_corners to numpy then rotate 45 degrees and stretch it by 0.8 then reconvert to list
s_shape_corners_np = np.array(corners)
theta = np.radians(45)  # rotate 45 degrees
rotation_matrix = np.array([[np.cos(theta), -np.sin(theta)],
                            [np.sin(theta), np.cos(theta)]])
s_shape_corners_rotated_np = s_shape_corners_np.dot(rotation_matrix)
s_shape_corners_stretched_np = s_shape_corners_rotated_np * 0.8  # stretch by 0.8 
s_shape_corners = s_shape_corners_stretched_np.tolist()

X, y = generate_polygon_data(corners=s_shape_corners, n_samples=n_samples, random_seed=random_seed)
print(f"Generated {len(X)} points for {shape_name}\n")
model, val_iou, nb_blocks = run_model_incremental(shape_name, X, y, max_blocks=max_blocks, n_sides=n_sides, random_seed=random_seed)

print(f"Final validation IoU for {shape_name} covered by {nb_blocks} x {n_sides}-sided blocks: {val_iou:.2%}")
print(model.summary())
````
